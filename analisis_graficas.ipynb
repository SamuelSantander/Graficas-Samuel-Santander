```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Creamos una base de datos con Pandas
datos = {
    "Carrera": ["Software", "Civil", "Ambiental", "Administración", "Medicina"],
    "Estudiantes": [145, 90, 60, 142, 110],
    "Satisfaccion": [4.5, 4.0, 3.8, 4.2, 4.7],
}
df = pd.DataFrame(datos)

# 2. Ponemos las 3 gráficas en una sola fila
fig, axs = plt.subplots(1, 3, figsize=(15, 4))

# --- GRÁFICA 1: Barras ---
axs[0].bar(df["Carrera"], df["Estudiantes"], color="teal")
axs[0].set_title("Cantidad de Estudiantes")

# --- GRÁFICA 2: Líneas ---
axs[1].plot(df["Carrera"], df["Estudiantes"], marker="o", color="orange")
axs[1].set_title("Tendencia por Carrera")

# --- GRÁFICA 3: Puntos ---
axs[2].scatter(df["Estudiantes"], df["Satisfaccion"], color="purple", s=100)
axs[2].set_title("Estudiantes vs Satisfacción")

# Ajustamos y guardamos la imagen.
plt.tight_layout()
plt.savefig("mis_tres_graficas.png")
plt.show()
```
